```python
import os
import chromadb
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
```

```python
# ==================== CONFIG ====================
EMBEDDING_MODEL = "keepitreal/vietnamese-sbert"
CHROMA_HOST = "localhost"
CHROMA_PORT = 8000
CHROMA_COLLECTION = "tax_documentation"
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name=EMBEDDING_MODEL,
    model_kwargs={"device": "cuda"}  # or "cpu"
)
```

```python
client = chromadb.HttpClient(
    host=CHROMA_HOST,
    port=CHROMA_PORT
)

vectorstore = Chroma(
    client=client,
    collection_name=CHROMA_COLLECTION,
    embedding_function=embeddings
)

collection = client.get_collection(name=CHROMA_COLLECTION)
```

```python
def print_vectorstore_info(collection):
    print("\n========== CHROMA VECTORSTORE INFO ==========")

    count = collection.count()
    print("Number of vectors:", count)

    docs = collection.get(limit=10, include=["documents"])
    ids = docs["ids"]

    print("Number of documents:", count)
    print("First 10 doc IDs:", ids)
    print("============================================\n")

print_vectorstore_info(collection)
```

```python
index = 0

results = collection.get(
    include=["documents", "metadatas", "embeddings"]
)

doc_id = results["ids"][index]
doc_text = results["documents"][index]
doc_metadata = results["metadatas"][index]
embedding_vector = results["embeddings"][index]

print(f"ID: {doc_id}")
print("Text:", doc_text)
print("Metadata:", doc_metadata)
print("Embedding length:", len(embedding_vector))
print("Embedding sample:", embedding_vector[:5])
```
