# 2) Splitting into sections

```python
lines = cleaned_text.splitlines()
```

```python
lines[:25]
```

```python
sections = []
current_section = {"title": None, "content": ""}

for line in lines:  # skip first 25 lines (assumed to be header)
    line = line.strip()
    if not line:
        continue  # skip empty lines

    # simple rule: line starts with a digit and a dot
    if line[0].isdigit() and "." in line.split()[0]:
        # if we already have a section, save it
        if current_section["title"]:
            sections.append(current_section)
        # start a new section
        current_section = {"title": line, "content": ""}
    else:
        # append to current section content
        current_section["content"] += " " + line

# add the last section
if current_section["title"]:
    sections.append(current_section)
```

# 3) Embedding document

```python
from transformers import AutoTokenizer, AutoModel
import torch
import numpy as np
from tqdm import tqdm
import faiss
```

```python
combined_sections = []

for s in sections:
    combined = s["title"] + "\n" + s["content"]
    combined_sections.append(combined.strip())
```

```python
combined_sections[2]
```

```python
# Load PhoBERT base model
tokenizer = AutoTokenizer.from_pretrained("vinai/phobert-base")
model = AutoModel.from_pretrained("vinai/phobert-base")

model.eval()  # Set model to evaluation mode
```

```python
def embed_text(text):
    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        padding=True,
        max_length=256
    )

    with torch.no_grad():
        outputs = model(**inputs)
        embeddings = outputs.last_hidden_state.mean(dim=1)  # average pooling

    return embeddings.squeeze().numpy()
```

```python
embeddings = []
for s in tqdm(combined_sections):
    embeddings.append(embed_text(s))
```

```python
convert_embeddings = np.array(embeddings).astype("float32")
```

```python
convert_embeddings.shape
```

```python
# Build FAISS index
index = faiss.IndexFlatL2(convert_embeddings.shape[1])
index.add(convert_embeddings)

def semantic_search(query, top_k=5):
    query_vec = embed_text(query)
    query_vec = np.array([query_vec]).astype("float32")  # ← make it 2D
    D, I = index.search(query_vec, top_k)
    return [combined_sections[i] for i in I[0]]
```

```python
# Example
query = "cơ sở tính thuế thu nhập cá nhân là gì ?"
results = semantic_search(query)
for r in results:
    print(r, "\n")
```
