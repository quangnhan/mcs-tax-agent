```python
import chromadb
from chromadb.config import Settings

# Connect to ChromaDB server at localhost:8000
try:
    client = chromadb.HttpClient(
        host="localhost",
        port=8000,
        settings=Settings(anonymized_telemetry=False)
    )
    print("✅ Connected to ChromaDB at localhost:8000")
except Exception as e:
    print(f"❌ Connection failed: {e}")
    print("Make sure ChromaDB Docker is running: docker-compose up -d")
    # Exit early if connection fails
    raise e
```

```python
# === Function to inspect ChromaDB collection ===
def print_vectorstore_info(collection_name):
    try:
        collection = client.get_collection(name=collection_name)
        print(f"\n========== CHROMA COLLECTION INFO ({collection_name}) ==========")
        print("Number of documents:", collection.count())
        print("Collection metadata:", collection.get().get('metadatas', []))
        print("First 10 document IDs:", collection.get(limit=10).get('ids', [])[:10])
        print("====================================================\n")
        return collection
    except Exception as e:
        print(f"Error inspecting collection {collection_name}: {e}")
        return None

# === Try Loading Existing Collection ===
collection = None
COLLECTION_NAME = "tax_documentation"  # Name for the ChromaDB collection

try:
    # Check if collection exists by trying to get it
    collection = client.get_collection(name=COLLECTION_NAME)
    print(f"Loaded existing ChromaDB collection: {COLLECTION_NAME}")
    print_vectorstore_info(COLLECTION_NAME)
except Exception as e:
    print(f"No existing collection '{COLLECTION_NAME}' found or error accessing it: {e}")
    print("Collection will be created when documents are added.")
```
